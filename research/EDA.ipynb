```python
def clean_data(df: pd.DataFrame) -> pd.DataFrame:
    """  
    Remove unnecessary columns safely.
    Parameters:
        df (pd.DataFrame): Input dataset
    Returns:
        pd.DataFrame: Cleaned dataset  s
    """
    if not isinstance(df, pd.DataFrame):
        raise TypeError("Input must be a pandas DataFrame")

    required_cols = ['Unnamed: 0', 'Booking_ID']
    return df.drop(columns=[c for c in required_cols if c in df.columns])   

#data.drop(columns=['Unnamed: 0', 'Booking_ID'] , inplace=True)
clean_data(data)
```

```python
#data.shape
class DataValidationError(Exception):
    pass
def validate_dataframe(df:pd.DataFrame) -> bool:
    if not isinstance(df, pd.DataFrame):
        raise TypeError("Input must be a pandas DataFrame")

    if df.empty: # df.shape[0] ==0
        raise ValueError("DataFrame is empty or Dataset is empty")
    if df.shape[0] < 1000:
        raise DataValidationError("Not enough data for training")
    if df.shape[1] < 5:
        raise ValueError("Insufficient features")
    
    return True

```

```python
validate_dataframe(data)
```

```python
# data.isnull().sum()  replaced  
def validate_missing_values(df :pd.DataFrame, threshold=0.05):
    missing_ratio = df.isnull().mean()
    
    if (missing_ratio > threshold).any():
        raise ValueError("Too many missing values detected")
validate_dataframe(data)
```

```python
#data.duplicated().sum()
def check_duplicates(df:pd.DataFrame):
    dup_count = df.duplicated().sum()
    if dup_count > 0:
        raise ValueError(f"Dataset contains {dup_count} duplicate rows")
check_duplicates(data)
```
